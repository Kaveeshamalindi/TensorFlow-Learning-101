### Import Data

```python
import pandas as pd
url="https://raw.githubusercontent.com/selva86/datasets/refs/heads/master/Ionosphere.csv"
df=pd.read_csv(url)
df.head()
```

```python
print(df.columns)
```

```python
x = df.drop(['Class'], axis=1)  # Independent Variables
y = df['Class']               # Dependent Variable
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2) #If there is a  100 raws, test=>20 raws & train=>80
```

### Import Dependencies

```python
from keras.models import Sequential, load_model
from keras.layers import Dense,Dropout
```

### Build ANN (Artificial Neural Network) Model

```python
model=Sequential()
model.add(Dense(units=10,activation='relu',input_dim=len(x_train.columns))) #1st hidden layer
model.add(Dense(units=5,activation='relu')) #1st hidden layer
model.add(Dense(units=1,activation='sigmoid',input_dim=len(x_train.columns))) #Output layer
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='sgd',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train, validation_data=(x_test,y_test), epochs=100, batch_size=32) #epoch=>No of times should be train #Batch size=> how many no of samples
```

```python
model.save('weights.h5')
```
